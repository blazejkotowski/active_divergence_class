Here, we will explore training-time active divergence techniques. 

1. Dataset mixing. We will train a DDSP model on a mixture of to datasets with different timbre characteristics to find out if the model can learn to generate sounds that are in between the two datasets. We will use instrumental samples.
2. Inspiring set. We will build an intentionally limited DDSP model incapable of modeling the dataset precisely, trying to reproduce animal vocalisations. Reproducing the paper "Modeling Animal Vocalizations through Synthesizers". This is implemented using torchsynth.
