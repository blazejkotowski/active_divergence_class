Experimenting with a simple harmonic + filtered noise DDSP model, we will explore various inference-time active divergence techniques to generate out-of-distribution samples.

To cover:
1. Out of distribution parameter sampling / Latent space exploration. Generating audio from parameters sampled from a distribution that is different from the training data.
2. Synthesis chain hacking. Modifying the synthesis chain of the model to create new sounds that are not present in the training data. Specifically, we will explore amplitude rolling, component limiting, and FIR-filter coefficient manipulation for the noise components of the model.
