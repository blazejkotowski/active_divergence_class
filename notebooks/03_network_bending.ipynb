Here, working with RAVE models we will explore model hacking approaches to generate out-of-distribution samples.

1. Activation-level Network Bending. We will be modifying the intermediary activations during the inference process. Basically, we need to intercept the generation, plug in between layers, modify the activations, and then let the generation process continue. We will explore various modifications to the activations, such as scaling, shifting, adding noise, or even replacing them with activations from other samples.
2. Weights-level Network Bending. This could be classified as model rewriting. We will explore various modifications to the weights in specific layers of the model, such as scaling, shifting, adding noise. We will see how this affects the generated audio.
3. Model blending. We will blend two different RAVE models trained on different datasets by interpolating their weights. We will explore how the blending ratio affects the generated audio and whether it can produce interesting out-of-distribution samples.
